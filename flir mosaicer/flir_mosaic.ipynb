```python
# imports

import numpy as np 
from skimage.transform import warp
import glob
import sys
import cv2
import matplotlib.pyplot as plt 
import spectral.io.envi as envi
import rasterio
from envi_header import find_hdr_file,read_hdr_file,write_envi_header
```

```python
# import images
flight = 'radiance/'  
filenames =  glob.glob(flight+"/*.img")
filenames.sort()

headers = glob.glob(flight + '/*.hdr')
headers.sort()

images = []

for i in range(len(filenames)):
    img = filenames[i]
    head = headers[i]

    imname = str(img)
    
    img = envi.open(head,img)
    img = img.open_memmap(writable=True)
    img = np.ma.array(img,mask=img==0)
    img = img.astype(np.float32)
    stack_img = np.dstack([img, img, img])
    images.append(stack_img)

print(len(images))
```

```python
#opencv method
stitcher = cv2.Stitcher_create()
stitcher.setPanoConfidenceThresh(0.0)
(status, stitched) = stitcher.stitch(images)

print(stitched)
```

```python
cv2.imwrite("stitched",stitched)
cv2.imshow("Stitched", stitched)
```
