```python
import numpy as np
import matplotlib.pyplot as plt

x, y = np.loadtxt('water_emis.txt',unpack=True,skiprows=3)
x_plot = []
y_plot = []

for item in x:
    x_plot.append(float(item))

for item in y:
    y_plot.append(float(item))
plt.title('water reflectance spectra')
plt.xlabel('wavelength[micron]')
plt.ylabel('reflectance [w/m^-2/sr^-2]')
plt.xlim([8, 14])
plt.grid()
plt.plot(x_plot,y_plot)
```
